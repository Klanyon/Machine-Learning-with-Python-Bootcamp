Q#1: Rice Market Analysis

```python
import numpy as np

# A rice seller recorded the following daily prices (LRD per bag)
prices = np.array([1200, 1250, 1180, 1300, 1220, 1270, 1240])

print('Prices:',         prices)
print('Mean Price:',     np.mean(prices).round(2))
print('Highest Price:',  np.max(prices).round(2))
print('Lowest Price:',   np.min(prices).round(2))
print('Dif. B/W the Highest & Lowest Prices',        np.max(prices).round(2) - np.min(prices).round(2))
print('It tells the seller quickly assess the baseline health of the business, though it should be cautions, as extreme highs or lows can skew the result.')

```

Q#2. Price Shock Investigation

```python
# Suppose the last price changes from 1240 LRD to 2000 LRD
prices = np.array([1200, 1250, 1180, 1300, 1220, 1240, 2000])
print('Prices:',         prices)
print('Mean Price:',     np.mean(prices).round(2))
print('The Old Mean is 1,237.14 and the New Mean is 1,341.43')
print('Std Dev:',        np.std(prices).round(2))
```

Q#5. Visualizing School Attendance

```python
# Attendance data:[35, 37, 40, 38, 42, 39, 41]
import matplotlib.pyplot as plt

days = list(range(1, 8))
Attendance = [35, 37, 40, 38, 42, 39, 41]

plt.figure(figsize=(8, 4))
plt.plot(days, Attendance, marker='o', color='darkorange', linewidth=2)
plt.title('Visualizing School Attendance')
plt.xlabel('Day')   ;  plt.ylabel('Attendance')
plt.grid(True)      ;  plt.tight_layout()   ;    plt.show()
```
