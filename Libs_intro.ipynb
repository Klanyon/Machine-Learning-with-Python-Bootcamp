MATPLOTLIB: CODE BLOCK 3.2

```python
import matplotlib.pyplot as plt

locations = ['Gobachop', 'Red Light', 'Kakata', 'Gbarnga']
avg_prices = [875, 910, 840, 860]

plt.figure(figsize = (7, 4))
plt.bar(locations, avg_prices, color = ['Red', 'Blue', 'Green' 'Yellow'])
plt.title('Average Gari price by Market (LIB)')
plt.ylabel('Price per Bag (LIB)') ; plt.ylim(800, 950)
plt.tight_layout() ; plt.show()
```
